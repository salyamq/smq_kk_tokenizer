```python
from datasets import load_dataset


dataset_name = "salyamq/kk-corpus-v1"
token = ""

ds = load_dataset(dataset_name, split="train", token = token)
```

```python
from tqdm import tqdm

texts = ds["text"]
n = len(texts)
print(f"total doc: {n}")

with open("full_corpus.txt", "w", encoding="utf-8") as f:
    for text in tqdm(texts):
        line = text.strip()
        if line:
            f.write(line + "\n")
```

```python
import sentencepiece as spm

spm.SentencePieceTrainer.train(
    input="full_corpus.txt",
    model_prefix="smq_kk_tokenizer",
    vocab_size=32000,
    model_type="bpe",
    character_coverage=0.9995,
    byte_fallback=True,
    normalization_rule_name="nmt_nfkc",
    remove_extra_whitespaces=True,
    split_by_whitespace=True,
    split_by_unicode_script=True,
    split_by_number=True,
    split_digits=True,
    unk_id=0, unk_piece="[UNK]",
    pad_id=1, pad_piece="[PAD]",
    bos_id=2, bos_piece="[BOS]",
    eos_id=3, eos_piece="[EOS]",
    user_defined_symbols=["[CLS]", "[SEP]", "[MASK]"],
    input_sentence_size=0,
    shuffle_input_sentence=True,
    train_extremely_large_corpus=True,
    max_sentencepiece_length=16,
    num_threads=80,   
    minloglevel=1,
)
```

```python
import sentencepiece as spm

sp = spm.SentencePieceProcessor(model_file="smq_kk_tokenizer.model")

print("vocab size:", sp.get_piece_size())

test_sentences = [
    "Қазақстан Республикасының астанасы — Астана қаласы.",
    "Мен биыл университетті бітіремін.",
    "2024 жылы 15 наурызда кездесу өтеді.",
]

for s in test_sentences:
    pieces = sp.encode(s, out_type=str)
    ids = sp.encode(s, out_type=int)
    decoded = sp.decode(ids)
    print(f"text: {s}")
    print(f"tokens: {pieces}")
    print(f"q: {len(pieces)}")
    print(f"decode == original: {decoded.strip() == s.strip()}")
    print("-" * 50)
```

```python
for tok in ["[UNK]", "[PAD]", "[BOS]", "[EOS]", "[CLS]", "[SEP]", "[MASK]"]:
    print(tok, "-", sp.piece_to_id(tok))
```
