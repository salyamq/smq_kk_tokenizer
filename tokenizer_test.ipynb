```python
import sentencepiece as spm

sp = spm.SentencePieceProcessor()
sp.load('smq_kk_tokenizer.model')

with open('chunk_099.txt', 'r', encoding='utf-8') as f:
    text = f.read()

print(f"chars: {len(text)}")
print(f"vocab size: {sp.get_piece_size()}") # 32k
```

```python
tokens = sp.encode(text, out_type=int)
pieces = sp.encode(text, out_type=str)
words = text.split()

n_tokens = len(tokens)
n_chars = len(text)
n_words = len(words)
n_bytes = len(text.encode('utf-8'))

unk_id = sp.unk_id()
n_unk = sum(1 for t in tokens if t == unk_id)

print(f"tokens: {n_tokens}")
print(f"words: {n_words}")
print(f"utf-8: {n_bytes}")
print(f"---" * 50)
print(f"Chars/token: {n_chars / n_tokens:.3f}")
print(f"Bytes/token: {n_bytes / n_tokens:.3f}")
print(f"Tokens/word:  {n_tokens / n_words:.3f}")
print(f"UNK tokens: {n_unk} ({100*n_unk/n_tokens:.2f}%)")
```

```python
from collections import Counter

lengths = [len(p) for p in pieces]
avg_len = sum(lengths) / len(lengths)

unique_pieces = set(pieces)
vocab_used_ratio = len(unique_pieces) / sp.get_piece_size()

print(f"avg length: {avg_len}")
print(f"unique pieces in the text: {len(unique_pieces)}")
print(f"using vocab {100*vocab_used_ratio:.2f}%")

counter = Counter(pieces)
print("top20 tokens:")
for piece, cnt in counter.most_common(20):
    print(f"  {piece!r}: {cnt}")
```
